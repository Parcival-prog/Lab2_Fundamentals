```python
import random

SEED_NUM = 7 
random.seed(SEED_NUM)

# Step 1: Generate a fixed number of reading
readings = [45.2, "ERROR", 78.9, -12, "N/A", 102.5] 

# Step 2: Validate each reading
valid_readings = []
for r in readings:
    try:
        value = float(r)
        valid_readings.append(value)
        print(f"Valid reading: {value}")
    except (ValueError, TypeError):
        print(f"Invalid reading skipped: {r}")

# Step 3: Classify valid readings by range
for v in valid_readings:
    if v < 0:
        status = "LOW"
    elif v <= 100:
        status = "NORMAL"
    else:
        status = "HIGH"
    print(f"{v} -> {status}")

# Step 4: Summary
print(f"Total valid: {len(valid_readings)} / Total received: {len(readings)}")
```

```python
SURNAME = "Canales"
SEED_NUM = 7

raw_signal = f"{SURNAME}{SEED_NUM}"
print(f"Generated Signal: {raw_signal}")

processed_signal = raw_signal.strip().upper()
print(f"Processed Signal: {processed_signal}")

print("Character Analysis")
letters = 0
digits = 0
spaces = 0
symbols = 0

for ch in processed_signal:
    if ch.isalpha():
        kind = "Letter"
        letters += 1
    elif ch.isdigit():
        kind = "Digit"
        digits += 1
    elif ch.isspace():
        kind = "Space"
        spaces += 1
    else:
        kind = "Symbol"
        symbols += 1
    print(f"'{ch}' -> {kind} (ASCII: {ord(ch)})")

if digits > letters:
    classification = "NUMERIC-DOMINANT"
elif letters > digits:
    classification = "TEXT-DOMINANT"
else:
    classification = "BALANCED"

print("--- Final Output ---")
print(f"Signal Classification: {classification}")
print(f"Letters: {letters}, Digits: {digits}, Spaces: {spaces}, Symbols: {symbols}")
print(f"Total Characters Processed: {len(processed_signal)}")

```

```python
SURNAME = "Canales"
SEED_NUM = 7

generated_password = f"{SURNAME.upper()}{SEED_NUM}"
attempt_limit = 3

print(f"Generated Password: {generated_password}")
print(f"Attempt Limit: {attempt_limit}")

attempts_made = 0
access_granted = False

while attempts_made < attempt_limit:
    pwd = input("Enter password: ")
    attempts_made += 1
    if pwd == generated_password:
        access_granted = True
        print("ACCESS GRANTED")
        break
    else:
        remaining = attempt_limit - attempts_made
        print(f"Access Denied. Attempts Remaining: {remaining}")

if not access_granted:
    print("SYSTEM LOCKED. Too many failed attempts.")

access_result = 'GRANTED'
final_state = 'UNLOCKED!!'

print("Final Output")
print(f"Attempts Made: {attempts_made}")
print(f"Access Result: {access_result}")
print(f"Final System State: {final_state}")

```
